# Task - 3 Model Creation #

### 1. Load the preprocessed dataset ###

```python
import pandas as pd

df = pd.read_csv("insurance_fraud_cleaned.csv")

print("Dataset shape:", df.shape)
print("\nTarget column:")
print(df['fraud reported'].value_counts())

print("\nTarget data type:")
print(df['fraud reported'].dtype)
```

### 2. Separate features and target ###

```python
X = df.drop('fraud reported', axis=1)
y = df['fraud reported']

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
```

### 3. Split data into training and testing sets ###

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

### 4. Create the Random Forest Classifier ###

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'
)

print("Random Forest Classifier created successfully!")
print(model)
```

### 5. Train the Random Forest Classifier ###

```python
model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

### 6. Generate predictions on the test data ###

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
print("Number of predictions:", len(y_pred))
print("\nFirst 20 predictions:")
print(y_pred[:20])
```

### 7. Save trained model ###

```python
import joblib

joblib.dump(model, "random_forest_model2.pkl")

print("Trained Random Forest model saved successfully!")
```
